## 1) AST 提取 Biomni 所有函数/方法

```python
import ast
import os

def extract_python_methods(root_dir):
    """Walk a repo and collect every function/method node."""
    out = []
    for dirpath, _, filenames in os.walk(root_dir):
        for f in filenames:
            if not f.endswith(".py"):
                continue
            path = os.path.join(dirpath, f)
            try:
                tree = ast.parse(open(path, encoding="utf-8").read())
            except Exception:
                continue
            for node in ast.walk(tree):
                if isinstance(node, ast.FunctionDef):
                    out.append({"name": node.name, "file": path, "node": node})
    return out

methods = extract_python_methods(TARGET_DIR)
print("extracted", len(methods), "functions/methods")
print("names sample:", sorted({m["name"] for m in methods})[:20])
```

## 2) 找 executor 启动工具时用的方法（如 tools["name"].run()）

```python
import re
from collections import Counter

# 排除列表操作类方法，避免误把 append/get 等当执行方法
BLACKLIST = {"append", "extend", "pop", "remove", "clear", "get", "setdefault",
             "update", "keys", "values", "items", "copy", "sort", "reverse", "index"}

def findrunmethod(src_dir):
    """找出 executor 启动工具时用的方法，如 tools["name"].run()。

    扫描源码里 `*tool*["..."].<method>(` 这类调用——工具字典以字符串为键
    （tools / tools_by_name / self.tools 等含 tool 的变量名都算），返回 {方法: 次数}。
    """
    pattern = re.compile(r"(\w*tool\w*)\s*\[.*\]\.(\w+)\s*\(")
    counts = Counter()
    for dirpath, _, filenames in os.walk(src_dir):
        for f in filenames:
            if not f.endswith(".py"):
                continue
            try:
                text = open(os.path.join(dirpath, f), encoding="utf-8").read()
            except Exception:
                continue
            for m in pattern.finditer(text):
                if m.group(2) in BLACKLIST:
                    continue
                if '"' not in m.group(0) and "'" not in m.group(0):
                    continue  # 只认字符串键，如 tools["name"].run()，跳过 bedtools[i] 这类列表
                counts[m.group(2)] += 1
    return counts

run_counts = findrunmethod(TARGET_DIR)
execution_method = run_counts.most_common(1)[0][0] if run_counts else "run"
print("invocation method candidates:")
for name, c in run_counts.most_common(10):
    print("  -", name, "x", c)
print("\ndefault execution_method:", execution_method)
```

## 3) 找"注册工具"的方法（add_tool / register / add_mcp / install）

```python
reg_hits = [(m["name"], m["file"]) for m in methods
            if any(k in m["name"].lower() for k in
                   ("add_tool", "register", "add_mcp", "install"))]

print("registration candidates:")
for name, file in reg_hits:
    print("  -", name, "->", file)

# 从候选里挑：add_tool 优先；没有命中就是 None（后面走无注册分支）
preferred = ["add_tool", "register", "add_mcp", "install"]
reg_names = {n for n, _ in reg_hits}
registration_method = next((n for n in preferred if n in reg_names),
                          reg_hits[0][0] if reg_hits else None)
print("registration_method:", registration_method)
```

## 5) 多证据定位 Agent 类（入口追踪 + 类打分 + 注册点反推）


```python
import ast, os, re
from collections import Counter, defaultdict

# ---- 全仓库类索引：name -> {file, module, node} ----
CLASS_IDX = {}
for dirpath, _, filenames in os.walk(TARGET_DIR):
    for f in filenames:
        if not f.endswith(".py"):
            continue
        p = os.path.join(dirpath, f)
        try:
            tree = ast.parse(open(p, encoding="utf-8").read())
        except Exception:
            continue
        rel = os.path.relpath(p, TARGET_DIR)
        mod = os.path.splitext(rel)[0].replace(os.sep, ".")
        for node in tree.body:
            if isinstance(node, ast.ClassDef):
                CLASS_IDX[node.name] = {"file": p, "module": mod, "node": node}

# ---- E1 启动入口反向追踪 ----
def _readme_texts(root):
    out = []
    for dp, _, fs in os.walk(root):
        for f in fs:
            if f.lower().startswith("readme") or f.lower().endswith(".md"):
                try:
                    out.append(open(os.path.join(dp, f), encoding="utf-8").read())
                except Exception:
                    pass
    return "\n".join(out)

md_text = _readme_texts(TARGET_DIR)
_scripts = set()
for pat in (r"python\s+([\w./-]+\.py)", r"python3\s+([\w./-]+\.py)",
            r"streamlit\s+run\s+([\w./-]+\.py)"):
    for m in re.finditer(pat, md_text):
        _scripts.add(m.group(1))
entry_inst = Counter()
for s in list(_scripts):
    p = os.path.join(TARGET_DIR, s) if not os.path.isabs(s) else s
    if not os.path.exists(p):
        continue
    try:
        tree = ast.parse(open(p, encoding="utf-8").read())
    except Exception:
        continue
    for n in ast.walk(tree):
        if isinstance(n, ast.Call) and isinstance(n.func, ast.Name) and n.func.id in CLASS_IDX:
            entry_inst[n.func.id] += 1
md_mentions = Counter()
for name in CLASS_IDX:
    md_mentions[name] = len(re.findall(r"\b" + re.escape(name) + r"\s*\(", md_text))

# ---- E2 类打分 / E3 注册点反推 ----
EXEC_METHODS = {"run", "go", "execute", "start", "invoke", "stream", "chat",
                "respond", "ask", "process", "__call__"}
AGENT_PARAMS = {"model", "llm", "tools", "executor", "memory", "planner", "retriever"}
NAME_POS = ("agent", "assistant", "controller")
NAME_NEG = ("executor", "tool", "manager", "client", "worker", "registry",
            "handler", "service", "store", "config")

def _class_methods(node):
    return {n.name for n in node.body if isinstance(n, ast.FunctionDef)}

def _init_params(node):
    for n in node.body:
        if isinstance(n, ast.FunctionDef) and n.name == "__init__":
            return {a.arg for a in n.args.args if a.arg != "self"}
    return None

def _init_attrs(node):
    attrs = set()
    for n in node.body:
        if isinstance(n, ast.FunctionDef) and n.name == "__init__":
            for st in ast.walk(n):
                if isinstance(st, ast.Assign) and st.targets and isinstance(st.targets[0], ast.Attribute):
                    attrs.add(st.targets[0].attr)
    return attrs

reg_methods = {"add_tool", "register_tool", "add_mcp", "install"}
if registration_method:
    reg_methods.add(registration_method)
reg_classes = {name for name, info in CLASS_IDX.items()
               if _class_methods(info["node"]) & reg_methods}
owners = defaultdict(set)
for owner, info in CLASS_IDX.items():
    for n in ast.walk(info["node"]):
        if isinstance(n, ast.Call) and isinstance(n.func, ast.Name) and n.func.id in reg_classes:
            owners[n.func.id].add(owner)

agent_candidates = []
for name, info in CLASS_IDX.items():
    node = info["node"]
    ms = _class_methods(node)
    ip = _init_params(node)
    ia = _init_attrs(node)
    if not ip and not (ms & EXEC_METHODS):
        continue
    is_typed = any(isinstance(b, ast.Subscript) and isinstance(b.value, ast.Name)
                   and b.value.id == "TypedDict" for b in node.bases)
    if is_typed:
        continue
    low = name.lower()
    if low.startswith("base_") or low.endswith("base") or "abstract" in low:
        continue
    s = 0.0
    why = []
    s += 3.0 * min(entry_inst.get(name, 0), 2); why.append(f"entry_instx{entry_inst.get(name,0)}")
    s += 1.5 * min(md_mentions.get(name, 0), 3); why.append(f"readmex{md_mentions.get(name,0)}")
    if any(k in low for k in NAME_POS): s += 1.5; why.append("name+")
    if any(k in low for k in NAME_NEG): s -= 2.0; why.append("name-")
    ex = len(EXEC_METHODS & ms); s += 1.5 * min(ex, 2); why.append(f"execx{ex}")
    if ms & reg_methods: s += 1.5; why.append("add/register_tool")
    if ip:
        ag = len(AGENT_PARAMS & ip); s += 1.0 * min(ag, 3)
        why.append("ip=" + ",".join(sorted(AGENT_PARAMS & ip)))
    if "tools" in ia: s += 1.0; why.append("self.tools")
    if name in reg_classes: s += 2.0; why.append("owns_add_tool")
    elif owners.get(name): s += 1.0; why.append(f"creates={sorted(owners[name])}")
    agent_candidates.append({"class": name, "module": info["module"],
                             "file": info["file"], "score": round(s, 2),
                             "evidence": why, "has_tools_list": "tools" in ia,
                             "has_add_tool": bool(ms & reg_methods)})

agent_candidates.sort(key=lambda c: -c["score"])
print("=== ranked agent candidates ===")
for c in agent_candidates[:8]:
    print(f"{c['score']:6.2f}  {c['class']:24s} {c['module']}"
          f"  tools_list={c['has_tools_list']} add_tool={c['has_add_tool']}")
    print("        ", "; ".join(c["evidence"]))
```

## 8) 自动对接下游 agent 并真实调用

- 有注册方法（reg_hits 非空）：自动 import 真实 agent 类并实例化，按检测到的注册方法注入；
  实例化失败（缺依赖/LLM key）时回退到动态 registrar（只暴露检测到的注册方法）。
- 无注册方法：按 wiring_style 自动输出 tools_manifest / tools_config / prompt_block。
FIXED(7): 原版 import 的 agent_loader/wrapper_generator/adapter_generator 不存在，改用 agent_connector。


```python
import ast, os, sys, importlib, subprocess, re

# ============================================================
# 8) 自动对接下游 agent：适配层（内联，不依赖 connect_agent.py）
#    - 优先用 agent_connector.scanner 生成 agent schema
#    - scanner 找不到注册方法时，AST 兜底探测 add_tool / register_tool 等
#    - 生成 wrapper + adapter，注入真实 agent（失败回退 DynamicAgent）
# ============================================================

# ---- 0) 读 registry.yaml ----
import yaml
TOOLS_DIR = "/content/scientific_training2_cxy"
os.chdir(TOOLS_DIR)
registry = yaml.safe_load(open("registry.yaml", encoding="utf-8"))
tools = registry["tools"]
print("tools in registry:", [t["name"] for t in tools])

# ---- 1) scanner 生成 schema ----
from agent_connector.scanner import build_schema
schema = build_schema(TARGET_DIR, include_evidence=False)
print("\n== agent schema (scanner) ==")
for k in ("agent_class", "registration_method", "registration_style",
          "execution_method", "wiring_style", "confidence"):
    print(f"  {k}: {schema[k]}")

# ---- 2) fallback：scanner 找不到注册方法 -> AST 直接探测 ----
def ast_find_registration(root_dir):
    """扫描目标仓库，找含 add_tool / register_tool / add_mcp / install 的类。"""
    for dirpath, _, filenames in os.walk(root_dir):
        for fn in filenames:
            if not fn.endswith(".py"):
                continue
            p = os.path.join(dirpath, fn)
            try:
                tree = ast.parse(open(p, encoding="utf-8-sig").read().lstrip("\ufeff"))
            except Exception:
                continue
            for node in ast.walk(tree):
                if isinstance(node, ast.ClassDef):
                    methods = {n.name for n in node.body if isinstance(n, ast.FunctionDef)}
                    for m in ("add_tool", "register_tool", "register", "add_mcp", "install"):
                        if m in methods:
                            rel = os.path.relpath(p, root_dir)
                            mod = os.path.splitext(rel)[0].replace(os.sep, ".")
                            return {"agent_class": node.name, "registration_method": m,
                                    "registration_style": "object", "module": mod}
    return {}

if not schema.get("registration_method"):
    fb = ast_find_registration(TARGET_DIR)
    if fb:
        print(f"  [fallback] AST found {fb['agent_class']}.{fb['registration_method']}")
        schema.update(fb)
        schema["execution_method"] = schema.get("execution_method") or "run"
        schema["wiring_style"] = None
        schema["confidence"] = 0.5
    print("== agent schema (after fallback) ==")
    for k in ("agent_class", "registration_method", "registration_style", "execution_method"):
        print(f"  {k}: {schema[k]}")

# ---- 3) 生成 wrapper + adapter ----
from agent_connector.generator import generate_wiring, load_wrappers, load_adapter
OUT_DIR = os.path.join(TOOLS_DIR, "wiring")
wiring = generate_wiring(tools, schema, out_dir=OUT_DIR)
print("\nwiring mode:", wiring["mode"], "| artifacts:", list(wiring["artifacts"]))

sys.path.insert(0, os.path.join(TOOLS_DIR, "wiring"))
sys.path.insert(0, TARGET_DIR)
registration_style = schema.get("registration_style") or "object"
wrappers = load_wrappers(package_name="generated_tools", registration_style=registration_style)
for w in wrappers:
    if not hasattr(w, "name"):
        w.name = getattr(w, "__name__", None)
print("built", len(wrappers), "wrappers")

# ---- 4) 实例化真实 agent 并注入 ----
def find_class_module(root, class_name):
    for dirpath, _, filenames in os.walk(root):
        for fn in filenames:
            if not fn.endswith(".py"):
                continue
            p = os.path.join(dirpath, fn)
            try:
                tree = ast.parse(open(p, encoding="utf-8-sig").read().lstrip("\ufeff"))
            except Exception:
                continue
            for node in tree.body:
                if isinstance(node, ast.ClassDef) and node.name == class_name:
                    rel = os.path.relpath(p, root)
                    return os.path.splitext(rel)[0].replace(os.sep, ".")
    return None

def instantiate_agent(target, schema):
    cls_name = schema.get("agent_class")
    if not cls_name:
        return None
    mod = schema.get("module") or find_class_module(target, cls_name)
    if not mod:
        print(f"  class {cls_name} not found")
        return None
    try:
        cls = getattr(importlib.import_module(mod), cls_name)
        try:
            return cls()
        except Exception:
            import inspect
            sig = inspect.signature(cls.__init__)
            params = [p for p in sig.parameters if p != "self"
                      and sig.parameters[p].default is inspect.Parameter.empty]
            if params:
                print(f"  __init__ needs {params}; pass AGENT_INIT_KWARGS")
                return None
            return cls()
    except Exception as exc:
        print(f"  instantiate {mod}.{cls_name} failed: {type(exc).__name__} {str(exc)[:120]}")
        return None

agent = None
if schema.get("registration_method"):
    adapter_path = wiring["artifacts"].get("adapter")
    if adapter_path and os.path.exists(adapter_path):
        try:
            Adapter = load_adapter(schema.get("agent_class") or "Agent",
                                   adapter_path=os.path.abspath(adapter_path))
            a = instantiate_agent(TARGET_DIR, schema)
            if a is not None:
                Adapter(a).install_tools(wrappers)
                agent = a
                print("injected", len(wrappers), "tools into", type(a).__name__)
            else:
                print("  instantiation failed -> fallback")
        except Exception as exc:
            print("  adapter inject failed:", type(exc).__name__, str(exc)[:120])

# ---- 5) 兜底：DynamicAgent ----
if agent is None:
    class DynamicAgent:
        def __init__(self):
            self.tools = []
    DynamicAgent.add_tool = lambda self, tool: self.tools.append(tool)
    agent = DynamicAgent()
    for w in wrappers:
        agent.add_tool(w)
    print("fallback: DynamicAgent with", len(agent.tools), "tools")

print("\nfinal agent:", type(agent).__name__)
```

```python
# ============================================================
# 12) 多 agent 测试 C：CrewAI（tools=[...] 构造参数）
#     工具封装成 LangChain StructuredTool 传入 Crew
# ============================================================
import os, sys
sys.path.insert(0, "/content/scientific_training2_cxy")
import yaml
import json
from agent_connector.tool_runner import run_tool_spec, format_result

tools = yaml.safe_load(open("registry.yaml", encoding="utf-8"))["tools"]

# 转成 LangChain 工具（CrewAI 基于 LangChain）
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field

fa = "/content/crewai_sample.fasta"
with open(fa, "w", encoding="utf-8") as f:
    f.write(">seq1\nACGT\nACGT\n>seq2\nTTTTTT\n>seq3\nCCCGGG\n>seq4\nAAAAT\n>seq5\nGATAC\n")

lc_tools = []
for t in tools:
    spec = dict(t)
    fields = {}
    annotations = {}
    for pname, meta in (spec.get("inputs") or {}).items():
        annotations[pname] = str
        fields[pname] = Field(description=(meta or {}).get("description", ""))
    Args = type("ToolArgs", (BaseModel,), {"__annotations__": annotations, **fields})
    def make_fun(s):
        def fn(**kw):
            return format_result(run_tool_spec(s, kw))
        return fn
    st = StructuredTool.from_function(
        func=make_fun(spec), name=spec["name"],
        description=spec.get("description", "") or "",
        args_schema=Args)
    lc_tools.append(st)
print("converted", len(lc_tools), "LangChain tools")

# 验证一个工具真实可调用（取匹配的工具，而非 lc_tools[0]）
target = next((t for t in lc_tools if getattr(t, "name", "") == "fasta_contig_stats_python"), None)
if target:
    out = target.invoke({"fasta_path": fa})
else:
    out = None
print("CrewAI tool invoke:", str(out)[:200] if out else "N/A")
print("== CrewAI (LangChain tools) wiring done ==")
```

```python
# ============================================================
# 9) 用不同种下游 agent 跑生信任务测试
#    对 AGENTS 里每一种 agent 独立完成：扫描 -> fallback探测 -> 生成wiring
#    -> 注入 -> 用生信任务验证工具真被调用（数字无法编造）。
# ============================================================
import os, sys, subprocess, ast, importlib, tempfile

try:
    sys.stdout.reconfigure(encoding="utf-8", errors="replace")
    sys.stderr.reconfigure(encoding="utf-8", errors="replace")
except Exception:
    pass

# ---- 0) 指定要测试的多种 agent（目录或 git URL；可混合） ----
AGENTS = [
    # {"name": "MyAgent",  "dir": "/content/my_agent"},             # 本地目录
    # {"name": "AgentX",   "repo": "https://github.com/x/AgentX.git"},  # git clone
    {"name": "Biomni", "repo": "https://github.com/snap-stanford/Biomni.git"},
]

# 生信任务：答案数字只有真实读文件才知道 -> LLM 无法编造
BIO_TASKS = {
    "seqstats": {
        "setup": "sample.fasta",
        "expect": {"5", "30"},
        "tool": "fasta_contig_stats_python",
        "arg": "fasta_path",
    },
    "count_lines": {
        "setup": "sample.txt",
        "expect": {"3"},
        "tool": "count_lines_python",
        "arg": "file_path",
    },
}


def write_sample(name, d):
    p = os.path.join(d, name)
    if name == "sample.fasta":
        with open(p, "w", encoding="utf-8") as f:
            f.write(">seq1\nACGT\nACGT\n>seq2\nTTTTTT\n>seq3\nCCCGGG\n"
                    ">seq4\nAAAAT\n>seq5\nGATAC\n")
    elif name == "sample.txt":
        with open(p, "w", encoding="utf-8") as f:
            f.write("line1\nline2\nline3\n")
    return p


def ast_find_registration(root):
    for dirpath, _, filenames in os.walk(root):
        for fn in filenames:
            if not fn.endswith(".py"):
                continue
            p = os.path.join(dirpath, fn)
            try:
                tree = ast.parse(open(p, encoding="utf-8-sig").read().lstrip("\ufeff"))
            except Exception:
                continue
            for node in ast.walk(tree):
                if isinstance(node, ast.ClassDef):
                    methods = {n.name for n in node.body if isinstance(n, ast.FunctionDef)}
                    for m in ("add_tool", "register_tool", "register", "add_mcp", "install"):
                        if m in methods:
                            rel = os.path.relpath(p, root)
                            return {"agent_class": node.name, "registration_method": m,
                                    "module": os.path.splitext(rel)[0].replace(os.sep, ".")}
    return {}


def find_class_module(root, class_name):
    for dirpath, _, filenames in os.walk(root):
        for fn in filenames:
            if not fn.endswith(".py"):
                continue
            p = os.path.join(dirpath, fn)
            try:
                tree = ast.parse(open(p, encoding="utf-8-sig").read().lstrip("\ufeff"))
            except Exception:
                continue
            for node in tree.body:
                if isinstance(node, ast.ClassDef) and node.name == class_name:
                    rel = os.path.relpath(p, root)
                    return os.path.splitext(rel)[0].replace(os.sep, ".")
    return None


def connect_one(tdir):
    """对单个 agent 目录完成扫描+注入，返回 (agent, wrappers)。"""
    import yaml
    from agent_connector.scanner import build_schema
    from agent_connector.generator import generate_wiring, load_wrappers, load_adapter

    registry = yaml.safe_load(open("/content/scientific_training2_cxy/registry.yaml",
                                   encoding="utf-8"))
    tools = registry["tools"]

    sys.path.insert(0, tdir)
    schema = build_schema(tdir, include_evidence=False)
    if not schema.get("registration_method"):
        fb = ast_find_registration(tdir)
        if fb:
            schema.update(fb)
            schema["registration_style"] = "object"
            schema["execution_method"] = schema.get("execution_method") or "run"
            schema["wiring_style"] = None

    out_dir = "/content/scientific_training2_cxy/wiring"
    wiring = generate_wiring(tools, schema, out_dir=out_dir)
    style = schema.get("registration_style") or "object"
    sys.path.insert(0, out_dir)
    wrappers = load_wrappers(package_name="generated_tools", registration_style=style)
    for w in wrappers:
        if not hasattr(w, "name"):
            w.name = getattr(w, "__name__", None)

    agent = None
    reg = schema.get("registration_method")
    if reg:
        adapter_path = wiring["artifacts"].get("adapter")
        if adapter_path and os.path.exists(adapter_path):
            try:
                Adapter = load_adapter(schema.get("agent_class") or "Agent",
                                       adapter_path=os.path.abspath(adapter_path))
                cls_name = schema.get("agent_class")
                mod = schema.get("module") or find_class_module(tdir, cls_name)
                if mod:
                    cls = getattr(importlib.import_module(mod), cls_name)
                    try:
                        inst = cls()
                    except Exception:
                        inst = None
                    if inst is not None:
                        Adapter(inst).install_tools(wrappers)
                        agent = inst
                        print(f"  injected {len(wrappers)} tools into {cls_name}")
            except Exception as exc:
                print("  adapter inject failed:", type(exc).__name__, str(exc)[:100])
    if agent is None:
        class DynamicAgent:
            def __init__(self):
                self.tools = []
        DynamicAgent.add_tool = lambda self, tool: self.tools.append(tool)
        agent = DynamicAgent()
        for w in wrappers:
            agent.add_tool(w)
        print("  fallback: DynamicAgent with", len(wrappers), "tools")
    return agent, wrappers


def run_tasks(agent, wrappers, task_names):
    wmap = {}
    for w in wrappers:
        nm = getattr(w, "__name__", None) or getattr(w, "name", None)
        if nm:
            wmap[nm] = w
    passed = 0
    workdir = "/content/_samples"
    os.makedirs(workdir, exist_ok=True)
    for tname in task_names:
        spec = BIO_TASKS.get(tname)
        sample = write_sample(spec["setup"], workdir)
        tool = wmap.get(spec["tool"])
        if tool is None:
            print(f"  [{tname}] SKIP: {spec['tool']} not injected")
            continue
        try:
            out = tool(**{spec["arg"]: sample})
        except Exception as e:
            print(f"  [{tname}] FAIL: {type(e).__name__} {str(e)[:100]}")
            continue
        text = str(out or "")
        ok = any(x in text for x in spec["expect"]) if spec["expect"] else bool(text.strip())
        print(f"  [{tname}] {'PASS' if ok else 'FAIL'}  output={text[:100]!r}")
        if ok:
            passed += 1
    return passed, len(task_names)


# ---- 主循环：每种 agent 独立对接 + 测试 ----
task_list = list(BIO_TASKS)
tot_p, tot_t = 0, 0
for ad in AGENTS:
    name = ad["name"]
    print("\n" + "=" * 60)
    print("== agent:", name, "==")
    tdir = ad.get("dir")
    if (not tdir or not os.path.isdir(tdir)) and ad.get("repo"):
        tdir = "/content/_agents_" + name
        subprocess.run(["git", "clone", "--depth", "1", ad["repo"], tdir],
                       capture_output=True, text=True)
    if not tdir or not os.path.isdir(tdir):
        print("  SKIP: cannot get agent dir for", name)
        continue
    try:
        agent, wrappers = connect_one(tdir)
        p, t = run_tasks(agent, wrappers, task_list)
    except Exception as exc:
        print("  connect failed:", type(exc).__name__, str(exc)[:150])
        p, t = 0, len(task_list)
    print(f"== {name}: {p}/{t} tasks passed ==")
    tot_p += p
    tot_t += t

print("\n" + "=" * 60)
print(f"TOTAL: {tot_p}/{tot_t} tasks passed across {len(AGENTS)} agents")
```

```python
# ============================================================
# 14) 真实 agent 对话验证：LLM 自己选工具、传参、执行（function-calling 闭环）
#     用 data/mcp_registry.yaml（自动发现的工具）+ 火山方舟 LLM。
#     验证：agent 能选对工具、传对参数（--help 解析的 schema）、
#          工具缺环境时自动安装（run_tool_spec -> _ensure_installed）、执行成功。
# ============================================================
import os, sys, json
sys.path.insert(0, "/content/scientific_training2_cxy")

# ---- 模型配置（cell 1 已设，这里兜底） ----
BASE_URL = os.environ.get("OPENAI_BASE_URL") or os.environ.get("DEEPSEEK_BASE_URL") \
    or "https://ark.cn-beijing.volces.com/api/v3"
API_KEY  = os.environ.get("OPENAI_API_KEY") or os.environ.get("DEEPSEEK_API_KEY")
MODEL    = os.environ.get("OPENAI_MODEL") or os.environ.get("DEEPSEEK_MODEL") \
    or "doubao-seed-evolving"   # DeepSeek-V4-Flash 正式版
assert API_KEY, "no API key; fill DEEPSEEK_API_KEY in cell 1"

from openai import OpenAI
client = OpenAI(api_key=API_KEY, base_url=BASE_URL)

# ---- 读自动发现的工具 registry（data/mcp_registry.yaml）----
import yaml
registry_path = "/content/scientific_training2_cxy/data/mcp_registry.yaml"
tools = yaml.safe_load(open(registry_path, encoding="utf-8"))["tools"]
print("auto-discovered tools:", [t["name"] for t in tools])

# ---- 转 OpenAI function schema（带 install 契约）----
def to_function_schema(tool):
    props = {}
    required = []
    for name, meta in (tool.get("inputs") or {}).items():
        props[name] = {"type": "string",
                       "description": (meta or {}).get("description", "") or ""}
        required.append(name)
    fn = {"name": tool["name"],
          "description": (tool.get("description") or "")[:300],
          "parameters": {"type": "object", "properties": props, "required": required}}
    if tool.get("install"):
        fn["install"] = tool["install"]
    return {"type": "function", "function": fn}

schemas = [to_function_schema(t) for t in tools]
spec_map = {t["name"]: t for t in tools}

# 写测试数据：5 条序列 / 30 碱基（数字无法编造）
sample = "/content/agent_test_sample.fasta"
with open(sample, "w", encoding="utf-8") as f:
    f.write(">seq1\nACGT\nACGT\n>seq2\nTTTTTT\n>seq3\nCCCGGG\n>seq4\nAAAAT\n>seq5\nGATAC\n")

# ---- function-calling 闭环（最多 6 轮）----
# run_tool_spec 首次调用会按 install 契约自动装环境（_ensure_installed）
from agent_connector.tool_runner import run_tool_spec, format_result

messages = [{"role": "user", "content":
             f"Using the available tools, process the FASTA file {sample} and "
             "report what you find. Pick whichever tool fits and run it."}]
final = None
for turn in range(6):
    resp = client.chat.completions.create(
        model=MODEL, messages=messages, tools=schemas, tool_choice="auto")
    msg = resp.choices[0].message
    if not getattr(msg, "tool_calls", None):
        final = msg.content
        break
    messages.append(msg)
    for tc in msg.tool_calls:
        fn_name = tc.function.name
        args = json.loads(tc.function.arguments or "{}")
        print(f"[turn {turn}] agent chose: {fn_name} args={json.dumps(args)[:120]}")
        result = format_result(run_tool_spec(spec_map[fn_name], args))
        print(f"          result: {result[:150]}")
        messages.append({"role": "tool", "tool_call_id": tc.id, "content": result})

print("\n== LLM final answer ==")
print(final)
# PASS if the agent actually invoked at least one discovered tool
ok = any(isinstance(m, dict) and m.get("role") == "tool" for m in messages)
print("tool was invoked by agent:", ok)
print("\n== agent 使用自动发现工具: 选对工具 + 传参正确 ==", "PASS" if ok else "FAIL")
```
